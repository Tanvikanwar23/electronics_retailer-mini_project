```python
%sql
CREATE SCHEMA IF NOT EXISTS electronics_retailer_clg.bronze;
```

```python
 # Read CSV from CUSTOMERS
 
df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("s3a://databricks-project-jman/electronics_retailer/Customers.csv")


df = df.toDF(*[col.replace(" ", "_") for col in df.columns])
df.display()
# Unity Catalog details
catalog_name = "electronics_retailer_clg"
schema_name = "bronze"
table_name = "customers"


# Save as table
df.write.mode("overwrite").saveAsTable(f"{catalog_name}.{schema_name}.{table_name}")


print(f"Table created: {catalog_name}.{schema_name}.{table_name}")
 
 
```
