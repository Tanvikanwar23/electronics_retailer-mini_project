```python
from datetime import datetime, timedelta
from pyspark.sql.functions import col

# ================================
# 1. GENERATE DATE RANGE
# ================================

start_date = datetime(2016, 1, 1)
end_date = datetime(2021, 12, 31)

date_list = [(start_date + timedelta(days=x),) 
             for x in range((end_date - start_date).days + 1)]

# ================================
# 2. CREATE DATAFRAME
# ================================

date_df = spark.createDataFrame(date_list, ["date"])

# ================================
# 3. ENSURE PROPER DATE TYPE
# ================================

date_dim = date_df.select(col("date").cast("date"))

# ================================
# 4. WRITE TO GOLD
# ================================

date_dim.write.format("delta") \
    .mode("overwrite") \
    .saveAsTable("electronics_retailer_clg.gold.dim_dates")

# ================================
# 5. CHECK OUTPUT
# ================================

display(date_dim)
date_dim.printSchema()

print("dim_dates created with only date column")
```
